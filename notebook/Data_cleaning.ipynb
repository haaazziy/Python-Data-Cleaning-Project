```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_report = missing_report[missing_report['Missing Values'] > 0]

missing_report
```

Standardizing Inconsistent Text Values

```python
text_columns = df.select_dtypes(include=['object']).columns

print("Text Columns:")
print(text_columns.tolist())
```

```python
for col in text_columns:
    print(f"\nColumn: {col}")
    print(df[col].value_counts().head(15))
```

```python
for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Unnecessary spaces removed!")
```

```python
for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.title() if isinstance(x, str) else x
    )

print("Text capitalization standardized!")
```

```python
for col in text_columns:
    print(f"\n{col}:")
    print(df[col].dropna().unique()[:10])
```

 Remove Unnecessary Spaces and Formatting Inconsistencies

```python
df.columns = df.columns.str.strip()

print("Column names standardized!")
print(df.columns.tolist())
```

```python
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()

print("Unnecessary spaces removed!")
```

```python
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True)

print("Multiple spaces standardized!")
```

```python
for col in df.select_dtypes(include='object').columns:
    print(f"\nColumn: {col}")
    print("Unique values:", df[col].nunique())
```
